```python
import os
import sys
import argparse
import pandas as pd
import matplotlib.pyplot as plt
import math
```

```python
df = pd.read_csv("/workdir/irk9/data/maps/maize/B73_v5_position.map", sep="\t", header=0)
```

```python
df.head()
```

```python
df["recomb_rate"] = df["Genetic map (cM)"] / df["Bin size (Mb)"]
df["Position(Mb)"] = df["Position"] / 1000000
```

```python
df[df["Chr"] == 1].sort_values("Position(Mb)").plot(x="Position(Mb)", y="recomb_rate", logy=True, xlabel="Position Mb",  ylabel="Recombination Rate (cM / bin size Mb)", title="Genetic Recombination Map: Chromosome 1")
```
