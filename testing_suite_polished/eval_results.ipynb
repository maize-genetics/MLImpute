```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df_maize = pd.read_csv('/workdir/irk9/data/phg-maize/eval_snp/results.tsv', sep='\t')
df_cassava = pd.read_csv('/workdir/irk9/data/phg-cassava/eval_snp/results.tsv', sep='\t')

df_maize_test2 = pd.read_csv('/workdir/irk9/data/phg-maize/test2/eval_snp/results.tsv', sep='\t')
df_cassava_test2 = pd.read_csv('/workdir/irk9/data/phg-cassava/test2_redo/eval_snp/results.tsv', sep='\t')

df_maize_test3 = pd.read_csv('/workdir/irk9/data/phg-maize/test3/eval_snp/results.tsv', sep='\t')
df_cassava_test3 = pd.read_csv('/workdir/irk9/data/phg-cassava/test3/eval_snp/results.tsv', sep='\t')
```

```python
df_maize = df_maize[df_maize["sample"] == "Oh7B"]
df_cassava = df_cassava[df_cassava["sample"] == "VEN25"]
df_maize = df_maize[df_maize["map"] == "nomap"]
df_cassava = df_cassava[df_cassava["map"] == "nomap"]
df_maize_test2 = df_maize_test2[df_maize_test2["map"] == "nomap"]
df_cassava_test2 = df_cassava_test2[df_cassava_test2["map"] == "nomap"]
df_maize_test3 = df_maize_test3[df_maize_test3["map"] == "nomap"]
df_cassava_test3 = df_cassava_test3[df_cassava_test3["map"] == "nomap"]

df_maize["partial_error_rate"] = 1 - df_maize["partial_allele_concordance"]
df_cassava["partial_error_rate"] = 1 - df_cassava["partial_allele_concordance"]
df_maize_test2["partial_error_rate"] = 1 - df_maize_test2["partial_allele_concordance"]
df_cassava_test2["partial_error_rate"] = 1 - df_cassava_test2["partial_allele_concordance"]
df_maize_test3["partial_error_rate"] = 1 - df_maize_test3["partial_allele_concordance"]
df_cassava_test3["partial_error_rate"] = 1 - df_cassava_test3["partial_allele_concordance"]
```
